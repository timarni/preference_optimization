### Ideas to clean/filter the synthetic Moove Example Dataset

1. Remove Invalid or Empty Entries
2. Standardize Formatting
3. Detect and Handle Duplicates
4. Remove Non-Medical or Out-of-Scope Questions
5. Normalize Case
6. Correct Typos and Grammar
7. Standardize Medical Terminology
8. Handle Sensitive Information
9. Validate Medical Context
10. Enforce Length Constraints
11. Convert to Structured Format (if necessary)
12. Validate Encoding

```python
import json
import re

def clean_jsonl_file(input_file, output_file):
    """
    Cleans a .jsonl file by:
    1. Removing invalid or empty entries.
    2. Standardizing formatting.
    12. Validating encoding (ensure UTF-8 compatibility).

    Args:
        input_file (str): Path to the input .jsonl file.
        output_file (str): Path to save the cleaned .jsonl file.
    """
    cleaned_entries = []

    try:
        # Read the .jsonl file line by line
        with open(input_file, "r", encoding="utf-8") as infile:
            for line in infile:
                try:
                    # Parse each line as JSON
                    entry = json.loads(line)
                    question = entry.get("question", "").strip()

                    # Step 1: Remove invalid or empty entries
                    if not question or question.lower() in {"` and ending with `", "`"}:
                        continue

                    # Step 2: Standardize formatting
                    question = re.sub(r"\*\*|\n", " ", question)  # Remove '**' and newlines
                    question = question.replace("\\u00b0", "°")   # Replace Unicode degrees
                    question = question.strip()                  # Remove leading/trailing spaces

                    # Add cleaned question to list
                    cleaned_entries.append({"question": question})
                except json.JSONDecodeError:
                    # Skip invalid JSON lines
                    continue

        # Step 12: Save cleaned data in UTF-8 encoding
        with open(output_file, "w", encoding="utf-8") as outfile:
            for entry in cleaned_entries:
                outfile.write(json.dumps(entry) + "\n")

        print(f"Cleaning complete. Cleaned data saved to {output_file}.")

    except Exception as e:
        print(f"An error occurred: {e}")

# Usage
input_path = "../resources/raw_generated_doctors_questions.jsonl"
output_path = "../resources/generated_doctors_questions.jsonl"

clean_jsonl_file(input_path, output_path)
```
